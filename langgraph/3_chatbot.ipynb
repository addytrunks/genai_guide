```python
from dotenv import load_dotenv
from langchain_groq import ChatGroq
from typing import TypedDict,Annotated
from langgraph.graph import StateGraph,START,END 
from langgraph.graph.message import add_messages

load_dotenv()
```

```python
llm = ChatGroq(model="llama-3.3-70b-versatile", temperature=0)

# https://docs.langchain.com/oss/python/langgraph/graph-api#reducers
class State(TypedDict):
    # Annotated preserves metadata for the messages field
    messages: Annotated[list,add_messages]
```

```python
def chatbot(state:State) -> State:
    return {"messages":[llm.invoke(state['messages'])]}

builder = StateGraph(State)
builder.add_node("chatbot_node", chatbot)

builder.add_edge(START, "chatbot_node")
builder.add_edge("chatbot_node", END)

graph = builder.compile()
```

```python
message = {"role":"user","content":"Who was the 16th president of the United States?"}
response = graph.invoke({"messages":[message]})
response['messages']
```

```python
state = None

while True:
    in_message = input("You: ")

    if in_message.lower() in ['exit','quit','q']:
        break

    if state is None:
        state:State = {"messages":{"role":"user","content":in_message}}
    else:
        state['messages'].append({"role":"user","content":in_message})
    state = graph.invoke(state)
    print("State:",state,"\n\n\n")
    out_message = state['messages'][-1].content
    print("Bot:",out_message)
```
